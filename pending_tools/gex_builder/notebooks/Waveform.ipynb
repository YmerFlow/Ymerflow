```python
time_shift_s = ref_wf.loc[zero_time_index].time_s
time_shift_s
```

```python
ref_wf['orig_time_s'] = ref_wf['time_s']
ref_wf['time_s'] = ref_wf['time_s'] - time_shift_s
ref_wf['n_wf_amp'] = ref_wf['t_current_a_'] / ref_wf['t_current_a_'].max()
ref_wf
```

```python
fig,ax = plt.subplots(1,1, sharex=True, figsize=(8, 8))
ax.plot(ref_wf.time_s, ref_wf.n_wf_amp)
ax.set_title('Xcalibur HeliTEM 21m Reference Waveform')
ax.set_ylabel('norm_current (A)')
plt.show()
```

```python
ref_wf.n_wf_amp[0:20]
```

```python
signs = np.sign(ref_wf.n_wf_amp)
# filter zeros betweeen two negs
for z_ind in signs[signs==0].index.to_list():
    if z_ind==0 and signs[z_ind+1]==-1:
        signs[z_ind] = -1
    elif signs[z_ind-1]==-1 and signs[z_ind+1]==-1:
        signs[z_ind] = -1
signs[0:20]
```

```python
sign_change = (signs * signs.shift(-1)).le(0)
sign_change[0:20]
```

```python
ref_wf.n_wf_amp[:11]
```

```python
first_crossing_index = sign_change[sign_change].index[0]
first_crossing_index
```

```python
new_ref_wf = ref_wf.iloc[first_crossing_index:].reset_index(drop=True)
new_ref_wf.loc[0, ['t_current_a_', 'n_wf_amp']] = [0,0]
new_ref_wf
```

```python
cut_time_index = new_ref_wf.loc[new_ref_wf.time_s==0].index[0]
cut_time_index
```
